```python
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
import re
import html
```

```python
df = pd.read_csv("data\export__14_08_26_111255(full_data).csv", encoding="cp1251", sep=";", 
                 usecols=['Id', 'Наименование', 'Краткое описание', 'Цена 5', 'URL'])

df = df.drop(df[df['Цена 5'] == 0].index)
```

```python
def clean_df(column):
    if pd.isna(column):
        return column
    
    if not isinstance(column, str):
        return column
    
    soup = BeautifulSoup(column, "lxml")
    
    clean = soup.get_text(" ", strip=True)
    clean = html.unescape(clean)
    clean = re.sub(r"/s+", " ", clean).strip()
    
    return clean
```

```python
for col in df.columns:
    df[col] = df[col].apply(clean_df)
```

```python
df.to_excel(r"data\cleaned_data.xlsx", index=False)
```
